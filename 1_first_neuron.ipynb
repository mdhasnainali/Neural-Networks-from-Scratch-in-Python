### A Layer of Neuron

```python
inputs = [1, 2, 3, 2.5]
weights1 = [0.2, 0.8, -0.5, 1]
weights2 = [0.5, -0.91, 0.26, -0.5]
weights3 = [-0.26, -0.27, 0.17, 0.87]
bias1 = 2
bias2 = 3
bias3 = 0.5
```

```python
outputs = [
    # Neuron 1:
    inputs[0]*weights1[0] +
    inputs[1]*weights1[1] +
    inputs[2]*weights1[2] +
    inputs[3]*weights1[3] + bias1,
    # Neuron 2:
    inputs[0]*weights2[0] +
    inputs[1]*weights2[1] +
    inputs[2]*weights2[2] +
    inputs[3]*weights2[3] + bias2,
    # Neuron 3:
    inputs[0]*weights3[0] +
    inputs[1]*weights3[1] +
    inputs[2]*weights3[2] +
    inputs[3]*weights3[3] + bias3]

outputs
```

```python
# More optimized way

inputs = [1, 2, 3, 2.5]
weights = [[0.2, 0.8, -0.5, 1],
            [0.5, -0.91, 0.26, -0.5],
            [-0.26, -0.27, 0.17, 0.87]]
biases = [2, 3, 0.5]
```

```python
# Output of current layer
layer_outputs = []
# For each neuron
for neuron_weights, neuron_bias in zip(weights, biases):
    # Zeroed output of given neuron
    neuron_output = 0
    # For each input and weight to the neuron
    for n_input, weight in zip(inputs, neuron_weights):
        # Multiply this input by associated weight
        # and add to the neuron's output variable
        neuron_output += n_input*weight
    # Add bias
    neuron_output += neuron_bias
    # Put neuron's result to the layer's output list
    layer_outputs.append(neuron_output)
print(layer_outputs)
```

### Tensors, Arrays and Vectors

**Array:** n array as an ordered homologous container.

```python
# Array Example

a = [  
    [1,2,3], 
    [4,5] 
]
# This is not homologous. So, it is not an array.

b = [
    [1,2,3],
    [4,5,6]
]
# This is homologous. So, it is an array.
```

**Matrix**: A matrix is a 2-dimensional array.

```python
# Matrix Example 

list_matrix_array = [[4, 2],
                    [5, 1],
                    [8, 2]]

# The first dimension is what’s inside the most outer brackets, and if we look at the above matrix, we can see 3 lists there: 
# [4,2], [5,1], and [8,2]; thus, the size in this dimension is 3 and each of those lists has to be the same shape to form 
# an array (and matrix in this case). The next dimension’s size is the number of elements inside this more inner pair of 
# brackets, and we see that it’s 2 as all of them contain 2 elements.
```

```python
list_matrix_array[0][1]
```

**Tensor**: A tensor object is an object that can be represented as an array

```python
# Tensor Example

# We couldn't give an object example but can give the representation example 

tolol = [[[1, 5, 6, 2],
        [3, 2, 1, 3]],
        [[5, 2, 1, 2],
        [6, 4, 8, 4]],
        [[2, 8, 5, 3],
        [1, 1,9,4]]]
# This is a 3D Array

# The first level of this array contains 3 matrices:
#           [[1,5,6,2],
#             [3,2,1,3]]
#      And
#           [[5,2,1,2],
#            [6,4,8,4]]
#      And
#           [[2,8,5,3],
#            [1,1,9,4]]

# That’s what’s inside the most outer brackets and the size of this dimension is then 3. If we look
# at the first matrix, we can see that it contains 2 lists — [1,5,6,2] and [3,2,1,3] so the
# size of this dimension is 2 — while each list of this inner matrix includes 4 elements. These 4 elements make up the 3rd and 
# last dimension of this matrix since there are no more inner brackets. Therefore, the shape of this array is (3, 2, 4) and it’s 
# a 3-dimensional array, since the shape contains 3 dimensions.
```

**Vector**: A vector is a list in Python or a 1-dimensional array in NumPy. This can be multi-dimensional

### A Single Neuron with NumPy

```python
import numpy as np

inputs = [1.0, 2.0, 3.0, 2.5]
weights = [0.2, 0.8, -0.5, 1.0]
bias = 2.0
```

```python
outputs = np.dot(weights, inputs) + bias
print(outputs)
```

### A Layer of Neurons with NumPy

```python
import numpy as np

inputs = [1.0, 2.0, 3.0, 2.5]
weights = [[0.2, 0.8, -0.5, 1],
            [0.5, -0.91, 0.26, -0.5],
            [-0.26, -0.27, 0.17, 0.87]]
biases = [2.0, 3.0, 0.5]
```

```python
layer_outputs = np.dot(weights, inputs) + biases
print(layer_outputs)
```

```python
# The np.dot() method treats the matrix as a list of vectors and performs a dot product of each of those vectors with the other 
# vector. In this example, we used that property to pass a matrix, which was a list of neuron weight vectors and a vector of 
# inputs and get a list of dot products — neuron outputs.
```

### A Batch of Data

To train, neural networks tend to receive data in batches. So far, the example input data have been
only one sample (or observation) of various features called a feature set.

Often, neural networks expect to take in many samples at a time for two reasons. One reason is that it’s faster to train in batches in parallel processing, and the other reason is that batches help with generalization during training. 

```python
# An example of a batch of data could look like:

batch = [[1, 2, 3, 2.5], 
        [2, 5, -1, 2], 
        [-1.5, 2.7, 3.3, -0.8]]

# This list of lists could be made into an array since it is homologous. 
# Note that each “list” in this larger list is a sample representing a feature set. 
# [1, 2, 3, 2.5], [2, 5, -1, 2], and [-1.5, 2.7, 3.3, -0.8] are all samples, 
# and are also referred to as feature set instances or observations.
```

```python
# In this example, we need to manage both matrices as lists of vectors and perform 
# dot products on all of them in all combinations, resulting in a list of lists of outputs, 
# or a matrix; this operation is called the matrix product.
```

### Matrix Product

The matrix product is an operation in which we have 2 matrices, and we are performing dot products of all combinations of rows from the first matrix and the columns of the 2nd matrix, resulting in a matrix of those atomic dot products

```python
# To perform a matrix product, the size of the second dimension of the left matrix 
# must match the size of the first dimension of the right matrix.
```

```python
a = [[1,2,3]]

b = [
    [1],
    [2],
    [3]
]

np.dot(a,b)
```

### Transposition for the Matrix Product

Transposition simply modifies a matrix in a way that its rows become columns and columns become rows.

Vector dot product need transpose: <br>
<img src="images/transpose.png" style="width:300px;height:150px;">

```python
a = [1,2,3]

row_expd = np.expand_dims(np.array(a), axis=0)     # Add the dimension(1) at the beginning. New dimension: 1x3
column_expd = np.expand_dims(np.array(a), axis=1)  # Add the dimension(1) at the ending. New dimension: 3x1

print("Row Expand:\n", row_expd)
print("Column Expand:\n", column_expd)
```

```python
# Transpose
import numpy as np

a = [1, 2, 3]
b = [2, 3, 4]
```

```python
a = np.array([a])
b = np.array([b]).T

print(np.dot(a, b))
```
